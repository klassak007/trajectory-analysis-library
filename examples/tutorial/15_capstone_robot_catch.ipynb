## 1. Read the logs

Each CSV is one attempt. The reader supplies the trial/sample roles, time
coordinate, and valid sequence lengths. Shorter recordings are padded; TAL keeps
track of which samples are real.

The trial catalog labels the controller and pairs the same six throws across
both controllers. These are **deterministic simulated logs**, with prescribed
controller offsets—not measured robot performance. Distances are in metres,
time is in seconds, and quaternion fields use **x, y, z, w** order.
See the [fixture description and generator](data/robot_catch/README.md).

```python
data_dir = Path("data/robot_catch")
if not data_dir.is_dir():
    data_dir = Path("examples/tutorial/data/robot_catch")
catalog = pd.read_csv(data_dir / "trials.csv").set_index("trial")
controller = xr.DataArray(catalog["controller"], dims="trial", name="controller")
ball_logs = read_csv_logs(
    str(data_dir / "camera/*.csv"), opts=CsvIngestOptions(time_col="time")
)
robot_logs = read_csv_logs(
    str(data_dir / "robot/*.csv"), opts=CsvIngestOptions(time_col="time")
)
ball_logs
```

## 2. Give the fields meaning

A Pose combines position and orientation. Declare the gripper's pose relative to
the camera, then register it with the graph. The ball's Position uses the same
camera frame. Field patterns spell out component order; TAL does not infer it.

The constructor declares the relationship. **`.register()` makes the Pose
available to graph queries.** No resolver callback is needed.

```python
graph = FrameGraph()
gripper = Pose.from_fields(
    robot_logs,
    position="gripper.position.{x,y,z}",
    rotation="gripper.rotation.{x,y,z,w}",
    parent="camera",
    child="gripper",
    graph=graph,
).register()
ball = Position.from_fields(
    ball_logs,
    "ball.position.{x,y,z}",
    parent="camera",
    child="ball",
    graph=graph,
)
gripper
```

## Inspect recording lengths and the matched population

The catalog preserves controller and throw identity. Camera recordings vary in length;
provider coverage must contain their valid query times. Paired throws have the same
camera observations but different prescribed gripper trajectories. Keep trial identity
visible before computing grouped summaries so population changes are detectable.

```python
display(catalog)
robot_sizes = ball_logs.as_dataset()["sequence_size"]
fig, ax = plt.subplots(figsize=(8, 3.5), layout="constrained")
for label, color, offset in [("predictive", BLUE, -0.12), ("reactive", ORANGE, 0.12)]:
    trials = catalog.index[catalog.controller == label]
    ax.scatter(
        robot_sizes.sel(trial=trials),
        catalog.loc[trials, "throw"] + offset,
        color=color,
        s=60,
        label=label,
    )
ax.set(
    yticks=range(1, 7),
    ylabel="Matched throw",
    xlabel="Valid camera observations",
    title="Pair identities before comparing controller outcomes",
)
ax.legend()
show(fig)
```

## 3. Make the gripper stand still

The registered pose is evaluated at the ball's timestamps. TAL interpolates the
translation and orientation, transforms every trial, and preserves each trial's
validity. The pose logs cover the entire camera recording.

```python
ball_in_gripper = ball.to_frame("gripper", graph=graph)
ball_in_gripper
```

### Follow one matched throw through the frame change

Throw 4 is recorded identically for both controller scenarios. The camera view therefore
shows one shared ball trajectory in gray and two prescribed gripper trajectories. In
gripper coordinates, the ball has a different relative trajectory for each controller.
Blue identifies predictive and orange reactive throughout this notebook.

The circle is a top-view projection of the 8 cm sphere. It helps locate horizontal
misses, but the next calculation uses all three position components. The representative
pair makes the transformation readable; all twelve attempts remain in the numerical analysis.

```python
palette = {"predictive": BLUE, "reactive": ORANGE}
capture_radius = 0.08
camera_view = ball.to_dataarray(copy="shallow")
local_view = ball_in_gripper.to_dataarray(copy="shallow")
gripper_view = gripper.as_dataset(copy="shallow")["position"]
illustrated_trials = catalog.index[catalog.throw == 4]
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2), layout="constrained")
shared_ball = camera_view.sel(trial=illustrated_trials[0])
axes[0].plot(
    shared_ball.sel(axis="x"),
    shared_ball.sel(axis="y"),
    color=GRAY,
    label="Shared ball: throw 4",
)
for trial in illustrated_trials:
    label = catalog.loc[trial, "controller"]
    g, r = gripper_view.sel(trial=trial), local_view.sel(trial=trial)
    axes[0].plot(
        g.sel(axis="x"),
        g.sel(axis="y"),
        ":",
        color=palette[label],
        label=f"{label} gripper",
    )
    axes[1].plot(r.sel(axis="x"), r.sel(axis="y"), color=palette[label], label=label)
axes[0].set(
    title="Camera: shared ball, different grippers",
    xlabel="Camera x (m)",
    ylabel="Camera y (m)",
)
axes[1].set(
    title="Gripper: relative paths separate",
    xlabel="Gripper x (m)",
    ylabel="Gripper y (m)",
    xlim=(-0.65, 0.5),
    ylim=(-0.35, 0.35),
)
axes[1].add_patch(Circle((0, 0), capture_radius, facecolor="#EDF1F4", edgecolor=GRAY))
axes[1].scatter([0], [0], marker="+", color=GRAY, s=90, label="gripper origin")
for ax in axes:
    ax.set_aspect("equal", adjustable="box")
    ax.grid()
    ax.legend(fontsize=9)
show(fig)
```

## 4. Compare all attempts

The vector norm gives the full 3-D separation. Reducing over `sample` produces
one closest sampled distance per attempt; grouping compares the controllers.
The labeled catalog is an external grouping key, so trial identities stay
explicit throughout.

`norm()` consumes the component axis and gives one distance per observation.
`min(dim="sample")` consumes the sequence and leaves one closest sampled distance per
trial. Grouping those summaries retains the explicit trial population. The mean of
trial minima is different from the minimum of a mean trajectory.

```python
distance = ball_in_gripper.norm().rename({"datavar": "distance"})
closest = distance.min(dim="sample")
comparison = closest.group.groupby(controller).mean(dim="trial")
comparison
```

## A top-view circle does not establish 3-D capture

The norm uses all three components. In this worked inset, (4, 4, 10) cm projects inside
the 8 cm circle but lies outside the sphere. The illustration explains why a planar
plot is only a projection. Reducing distance over sample selects the closest recorded
measurement and removes the sequence axis; it does not find a continuous minimum.

```python
sphere_example = Position.from_fields(
    AnalysisLayoutSpec().wrap(xr.Dataset({"p.x": 0.04, "p.y": 0.04, "p.z": 0.10})),
    "p.{x,y,z}",
)
sphere_distance = sphere_example.norm()
display(
    pd.DataFrame(
        {
            "measure": [
                "Top-view separation (cm)",
                "Full separation (cm)",
                "Capture radius (cm)",
            ],
            "value": [
                100 * np.hypot(0.04, 0.04),
                100 * float(sphere_distance.to_dataarray()),
                8.0,
            ],
        }
    )
)
fig, axes = plt.subplots(1, 2, figsize=(9, 4), layout="constrained")
for ax, xy, title, vertical in [
    (axes[0], [4, 4], "Top view: inside the projected circle", "y (cm)"),
    (axes[1], [4, 10], "Side view: outside the projected circle", "z (cm)"),
]:
    ax.add_patch(Circle((0, 0), 8, facecolor="#F1F4F7", edgecolor=GRAY))
    ax.scatter([0], [0], marker="+", color=GRAY, s=90)
    ax.scatter([xy[0]], [xy[1]], color=ORANGE, s=75)
    ax.plot([0, xy[0]], [0, xy[1]], ":", color=ORANGE)
    ax.annotate(str(tuple(xy)), xy, xytext=(7, 7), textcoords="offset points")
    ax.set(
        xlim=(-9, 12),
        ylim=(-9, 14),
        aspect="equal",
        xlabel="x (cm)",
        ylabel=vertical,
        title=title,
    )
show(fig)
```

A trial is **capture-eligible** if any observed valid sample is inside the
8 cm sphere. `events.mask` returns an xarray boolean array. Wrap the per-trial
indicator with its explicit batch layout to use the same TAL grouping operation.
This is a sampled geometric criterion; it does not model collisions, finger
closure, or prove a successful catch between samples.

```python
inside = distance.events.mask(distance < capture_radius)
eligible = AnalysisLayoutSpec(batch_dims=("trial",)).wrap(
    inside.any("sample").astype(float).rename("eligible"),
)
capture_rate = eligible.group.groupby(controller).mean(dim="trial")
scorecard = pd.DataFrame(
    {
        "Mean closest distance (cm)": 100
        * comparison.to_dataarray(copy="shallow").to_series(),
        "Capture-eligible (%)": 100
        * capture_rate.to_dataarray(copy="shallow").to_series(),
    }
).rename_axis("controller")
scorecard.round(1)
```

## Inspect each matched pair instead of connecting separate throws

Each horizontal connector below joins predictive and reactive results for the **same**
throw. Throw identity occupies its own row; neighboring rows are independent cases.
The vertical line marks the 8 cm radius. Points to its left have a closest sampled
distance below that threshold.

This view explains variation hidden by the group mean. The prescribed scenarios include
one easy reactive case and one predictive miss; controller labels alone do not determine
the geometric result.

```python
paired_closest = (
    closest.to_dataarray()
    .to_series()
    .rename("closest_m")
    .to_frame()
    .join(catalog)
    .pivot(index="throw", columns="controller", values="closest_m")
)
fig, ax = plt.subplots(figsize=(9, 4), layout="constrained")
ax.hlines(
    paired_closest.index,
    100 * paired_closest.predictive,
    100 * paired_closest.reactive,
    color=GRAY,
    alpha=0.5,
)
for label in ["predictive", "reactive"]:
    ax.scatter(
        100 * paired_closest[label],
        paired_closest.index,
        color=palette[label],
        s=65,
        label=label,
    )
ax.axvline(100 * capture_radius, color=GRAY, linestyle="--", label="8 cm radius")
ax.set(
    yticks=paired_closest.index,
    ylabel="Matched throw",
    xlabel="Closest sampled distance (cm)",
    title="One connector per matched throw",
)
ax.legend()
show(fig)
```

## 5. Line up the approaches, not the clocks

Align every attempt to its **first observed entry within 35 cm**. All twelve
attempts enter this larger approach zone, including those that miss the capture
zone. This avoids selecting only successful attempts for the comparison.

`tau = 0` is the sampled approach event, not closest approach or contact. Each
window includes 120 ms before and 320 ms after entry, at 10 ms spacing.

```python
approach = distance.events.around(
    distance < 0.35,
    edge="enter",
    pre=0.12,
    post=0.32,
    dt=0.01,
    layout="segments",
)
approach
```

The same controller colors now follow two time coordinates. The recording-time view
includes different arrival times; the event view uses `tau=0` at first observed entry
into the 35 cm approach zone. All twelve attempts remain in both views. The common
vertical scale and 8 cm reference line make geometric separation comparable.

An approach-zone event is useful precisely because it includes misses. Aligning on
capture-zone entry would select a different, smaller population.

```python
plot_options = AOVizOptions(
    by=("trial",),
    max_overlay_items=12,
    kwargs={
        "color": [palette[label] for label in catalog["controller"]],
        "legend": False,
        "width": 520,
        "height": 330,
        "ylim": (0.0, 0.8),
        "ylabel": "Separation (m)",
    },
)
clock_plot = distance.viz.line(opts=plot_options).relabel("Recording time")
event_plot = (
    approach.isel(event=0)
    .viz.line(
        opts=AOVizOptions(
            x="tau", by=("trial",), max_overlay_items=12, kwargs=plot_options.kwargs
        )
    )
    .relabel("Time from 35 cm entry")
)
threshold = hv.HLine(capture_radius).opts(
    color="#555555", line_dash="dashed", line_width=1
)
comparison_plot = ((clock_plot * threshold) + (event_plot * threshold)).cols(2)
display(comparison_plot)
```

## Worked comparison: capture radius and anchor population

Vary capture radius while holding the larger approach anchor fixed. Each radius is
evaluated through TAL's condition mask, then wrapped as an explicitly declared batch
indicator for grouped reduction. A capture-zone anchor would include only trials with
an observed entry into that smaller zone; it would remove most misses from a window
comparison. The 35 cm anchor retains all twelve approaches.

```python
radius_values = np.array([0.04, 0.08, 0.12])
radius_rates = []
for radius in radius_values:
    radius_mask = distance.events.mask(distance < float(radius))
    radius_indicator = AnalysisLayoutSpec(batch_dims=("trial",)).wrap(
        radius_mask.any("sample").astype(float).rename("eligible")
    )
    radius_summary = radius_indicator.group.groupby(controller).mean(dim="trial")
    radius_rates.append(radius_summary.to_dataarray().to_series())
radius_table = pd.DataFrame(radius_rates, index=radius_values)
display(radius_table)
capture_entries_demo = distance.events.at_boundaries(
    distance < capture_radius, opts=AtBoundariesOptions(edges="enter", mode="first")
)
capture_anchor_count = int(capture_entries_demo.as_dataset()["valid"].sum())
assert capture_anchor_count == 6
fig, ax = plt.subplots(figsize=(8, 3.5), layout="constrained")
for label in ["predictive", "reactive"]:
    ax.scatter(
        100 * radius_values,
        100 * radius_table[label],
        s=70,
        color=palette[label],
        label=label,
    )
    for radius, rate in zip(radius_values, radius_table[label], strict=True):
        ax.annotate(
            f"{round(6 * rate)}/6",
            (100 * radius, 100 * rate),
            xytext=(8, -12 if label == "predictive" else 9),
            textcoords="offset points",
            color=palette[label],
        )
ax.set(
    xticks=[4, 8, 12],
    ylim=(-8, 110),
    xlabel="Capture radius (cm)",
    ylabel="Eligible trials (%)",
    title="Three tested radius choices; six trials per controller",
)
ax.legend()
show(fig)
display(
    pd.DataFrame(
        {
            "anchor": ["35 cm approach", "8 cm capture"],
            "trials with observed entry": [12, capture_anchor_count],
        }
    )
)
```

Each dot is a separately evaluated radius choice, with its eligible count shown beside
it. The plot leaves intermediate radii unspecified rather than implying a linear change
between choices. The anchor table exposes a separate population decision: the larger
approach event retains twelve trials, while capture entry retains only six.

## What did we learn?

For these prescribed trajectories, **5 of 6 predictive attempts** enter the
capture sphere, versus **1 of 6 reactive attempts**. The gripper-frame view exposes
a larger lateral offset in the reactive examples. Event alignment makes the
approaches comparable even though their recording times differ.

The result follows from visible TAL operations: CSV ingestion → field assembly →
Pose registration → frame transformation → norm → reduction/grouping → event windows.
The raw-data generator performs none of those analysis steps.

The radius comparison is worked out above. The approach population stays fixed while the capture criterion changes, keeping misses visible. A change of anchor definition changes the population and therefore changes what the event-aligned comparison can establish.
