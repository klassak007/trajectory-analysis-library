```python
import numpy as np

from tal.core.event_ops import AtBoundariesOptions
```

## 1. Read the logs

Each CSV is one attempt. The reader supplies the trial/sample roles, time
coordinate, and valid sequence lengths. Shorter recordings are padded; TAL keeps
track of which samples are real.

The trial catalog labels the controller and pairs the same six throws across
both controllers. These are **deterministic simulated logs**, with prescribed
controller offsets—not measured robot performance. Distances are in metres,
time is in seconds, and quaternion fields use **x, y, z, w** order.
See the [fixture description and generator](data/robot_catch/README.md).

```python
data_dir = Path("data/robot_catch")
if not data_dir.is_dir():
    data_dir = Path("examples/tutorial/data/robot_catch")
catalog = pd.read_csv(data_dir / "trials.csv").set_index("trial")
controller = xr.DataArray(catalog["controller"], dims="trial", name="controller")
ball_logs = read_csv_logs(str(data_dir / "camera/*.csv"), opts=CsvIngestOptions(time_col="time"))
robot_logs = read_csv_logs(str(data_dir / "robot/*.csv"), opts=CsvIngestOptions(time_col="time"))
ball_logs
```

## 2. Give the fields meaning

A Pose combines position and orientation. Declare the gripper's pose relative to
the camera, then register it with the graph. The ball's Position uses the same
camera frame. Field patterns spell out component order; TAL does not infer it.

The constructor declares the relationship. **`.register()` makes the Pose
available to graph queries.** No resolver callback is needed.

```python
graph = FrameGraph()
gripper = Pose.from_fields(
    robot_logs, position="gripper.position.{x,y,z}", rotation="gripper.rotation.{x,y,z,w}",
    parent="camera", child="gripper", graph=graph,
).register()
ball = Position.from_fields(
    ball_logs, "ball.position.{x,y,z}", parent="camera", child="ball", graph=graph,
)
gripper
```

## Inspect recording lengths and the matched population

The catalog preserves controller and throw identity. Camera recordings vary in length;
provider coverage must contain their valid query times. Paired throws have the same
camera observations but different prescribed gripper trajectories. Keep trial identity
visible before computing grouped summaries so population changes are detectable.

```python
display(catalog)
robot_sizes = ball_logs.as_dataset()["sequence_size"]
fig, ax = plt.subplots(figsize=(7, 3.5), layout="constrained")
ax.bar(robot_sizes.trial.to_numpy(), robot_sizes.to_numpy(), color=["#087e8b" if v == "predictive" else "#cf4b32" for v in catalog.controller])
ax.tick_params(axis="x", rotation=60)
ax.set(ylabel="Valid camera samples", title="Matched throws retain their recording identity")
display(fig)
plt.close(fig)
```

## 3. Make the gripper stand still

The registered pose is evaluated at the ball's timestamps. TAL interpolates the
translation and orientation, transforms every trial, and preserves each trial's
validity. The pose logs cover the entire camera recording.

```python
ball_in_gripper = ball.to_frame("gripper", graph=graph)
ball_in_gripper
```

**The same data, seen from two frames.** Solid lines are ball trajectories;
dotted lines on the left are the moving gripper. On the right the gripper is
always at the origin. The shaded circle is the top-view projection of the
capture sphere. The numerical decision below uses all **three** dimensions.

This geometry view uses Matplotlib; the time-series views below use TAL's plot
accessor. Dataset exposure here is only for presentation.

```python
palette = {"predictive": "#087e8b", "reactive": "#cf4b32"}
capture_radius = 0.08
camera_view = ball.to_dataarray(copy="shallow")
local_view = ball_in_gripper.to_dataarray(copy="shallow")
gripper_view = gripper.as_dataset(copy="shallow")["position"]
fig, axes = plt.subplots(1, 2, figsize=(12, 4.6), layout="constrained")
for label, color in palette.items():
    trials = catalog.index[catalog["controller"] == label]
    p, g, r = (view.sel(trial=trials) for view in (camera_view, gripper_view, local_view))
    axes[0].plot(p.sel(axis="x").T, p.sel(axis="y").T, color=color, alpha=0.55)
    axes[0].plot(g.sel(axis="x").T, g.sel(axis="y").T, ":", color=color, alpha=0.45)
    axes[1].plot(r.sel(axis="x").T, r.sel(axis="y").T, color=color, alpha=0.8)
    axes[1].plot([], [], color=color, label=label)
axes[0].set(title="Camera frame · both objects move", xlabel="camera x (m)", ylabel="camera y (m)")
axes[1].set(title="Gripper frame · the misses separate", xlabel="gripper x (m)", ylabel="gripper y (m)",
            xlim=(-0.65, 0.5), ylim=(-0.35, 0.35))
axes[1].add_patch(Circle((0, 0), capture_radius, color="#087e8b", alpha=0.16))
axes[1].scatter(0, 0, color="#243447", marker="+", s=100, label="gripper origin")
for ax in axes:
    ax.set_aspect("equal", adjustable="box")
    ax.grid(alpha=0.18)
axes[1].legend(frameon=False, loc="upper left")
display(fig)
plt.close(fig)
```

## 4. Compare all attempts

The vector norm gives the full 3-D separation. Reducing over `sample` produces
one closest sampled distance per attempt; grouping compares the controllers.
The labeled catalog is an external grouping key, so trial identities stay
explicit throughout.

```python
distance = ball_in_gripper.norm().rename({"datavar": "distance"})
closest = distance.min(dim="sample")
comparison = closest.group.groupby(controller).mean(dim="trial")
comparison
```

## A top-view circle does not establish 3-D capture

The norm uses all three components. In this worked inset, (4, 4, 10) cm projects inside
the 8 cm circle but lies outside the sphere. The illustration explains why a planar
plot is only a projection. Reducing distance over sample selects the closest recorded
measurement and removes the sequence axis; it does not find a continuous minimum.

```python
sphere_example = Position.from_fields(AnalysisLayoutSpec().wrap(xr.Dataset({"p.x": 0.04, "p.y": 0.04, "p.z": 0.10})), "p.{x,y,z}")
sphere_distance = sphere_example.norm()
display(pd.DataFrame({"measure": ["Top-view separation (cm)", "Full separation (cm)", "Capture radius (cm)"],
    "value": [100*np.hypot(0.04, 0.04), 100*float(sphere_distance.to_dataarray()), 8.]}))
fig, ax = plt.subplots(figsize=(5, 3), layout="constrained")
ax.bar(["Top view", "Full 3-D"], [100*np.hypot(0.04, 0.04), 100*float(sphere_distance.to_dataarray())])
ax.axhline(8, color="grey", linestyle="--", label="Capture radius")
ax.set(ylabel="Separation (cm)")
ax.legend()
display(fig)
plt.close(fig)
```

A trial is **capture-eligible** if any observed valid sample is inside the
8 cm sphere. `events.mask` returns an xarray boolean array. Wrap the per-trial
indicator with its explicit batch layout to use the same TAL grouping operation.
This is a sampled geometric criterion; it does not model collisions, finger
closure, or prove a successful catch between samples.

```python
inside = distance.events.mask(distance < capture_radius)
eligible = AnalysisLayoutSpec(batch_dims=("trial",)).wrap(
    inside.any("sample").astype(float).rename("eligible"),
)
capture_rate = eligible.group.groupby(controller).mean(dim="trial")
scorecard = pd.DataFrame({
    "Mean closest distance (cm)": 100 * comparison.to_dataarray(copy="shallow").to_series(),
    "Capture-eligible (%)": 100 * capture_rate.to_dataarray(copy="shallow").to_series(),
}).rename_axis("controller")
scorecard.round(1)
```

## Inspect paired closest approaches

Mean closest distance and capture-eligible fraction summarize different aspects of the
same per-trial observations. Pair the six throws explicitly rather than pairing rows by
accidental position. The line plot below preserves throw identity and shows differences
hidden by a single group mean. These remain prescribed simulated scenarios.

```python
paired_closest = closest.to_dataarray().to_series().rename("closest_m").to_frame().join(catalog).pivot(index="throw", columns="controller", values="closest_m")
fig, ax = plt.subplots(figsize=(6, 3), layout="constrained")
for label in ["predictive", "reactive"]:
    ax.plot(paired_closest.index, 100*paired_closest[label], "o-", color=palette[label], label=label)
ax.axhline(100*capture_radius, color="grey", linestyle="--")
ax.set(xlabel="Matched throw", ylabel="Closest sampled distance (cm)")
ax.legend()
display(fig)
plt.close(fig)
```

## 5. Line up the approaches, not the clocks

Align every attempt to its **first observed entry within 35 cm**. All twelve
attempts enter this larger approach zone, including those that miss the capture
zone. This avoids selecting only successful attempts for the comparison.

`tau = 0` is the sampled approach event, not closest approach or contact. Each
window includes 120 ms before and 320 ms after entry, at 10 ms spacing.

```python
approach = distance.events.around(
    distance < 0.35, edge="enter", pre=0.12, post=0.32, dt=0.01, layout="segments",
)
approach
```

Same colours, same trials. The event view removes differences in arrival time;
the orange misses now separate from the teal approaches. The dashed horizontal
line is the capture radius. TAL's plot accessor uses the declared parameter and
validity, so no padding masks or plotting loops are needed here.

```python
plot_options = AOVizOptions(
    by=("trial",), max_overlay_items=12,
    kwargs={"color": [palette[label] for label in catalog["controller"]], "legend": False,
            "width": 520, "height": 310, "ylabel": "Separation (m)"},
)
clock_plot = distance.viz.line(opts=plot_options).relabel("Recording time")
event_plot = approach.isel(event=0).viz.line(opts=plot_options).relabel("Time from 35 cm entry")
threshold = hv.HLine(capture_radius).opts(color="#555555", line_dash="dashed", line_width=1)
comparison_plot = ((clock_plot * threshold) + (event_plot * threshold)).cols(2)
display(comparison_plot)
```

## Worked comparison: capture radius and anchor population

Vary capture radius while holding the larger approach anchor fixed. Each radius is
evaluated through TAL's condition mask, then wrapped as an explicitly declared batch
indicator for grouped reduction. A capture-zone anchor would include only trials with
an observed entry into that smaller zone; it would remove most misses from a window
comparison. The 35 cm anchor retains all twelve approaches.

```python
radius_values = np.array([0.04, 0.08, 0.12])
radius_rates = []
for radius in radius_values:
    radius_mask = distance.events.mask(distance < float(radius))
    radius_indicator = AnalysisLayoutSpec(batch_dims=("trial",)).wrap(radius_mask.any("sample").astype(float).rename("eligible"))
    radius_summary = radius_indicator.group.groupby(controller).mean(dim="trial")
    radius_rates.append(radius_summary.to_dataarray().to_series())
radius_table = pd.DataFrame(radius_rates, index=radius_values)
display(radius_table)
capture_entries_demo = distance.events.at_boundaries(distance < capture_radius, opts=AtBoundariesOptions(edges="enter", mode="first"))
capture_anchor_count = int(capture_entries_demo.as_dataset()["valid"].sum())
assert capture_anchor_count == 6
fig, ax = plt.subplots(figsize=(6, 3), layout="constrained")
for label in ["predictive", "reactive"]:
    ax.plot(100*radius_values, 100*radius_table[label], "o-", color=palette[label], label=label)
ax.set(xlabel="Capture radius (cm)", ylabel="Eligible trials (%)", title="Fixed trial population and approach anchor")
ax.legend()
display(fig)
plt.close(fig)
display(pd.DataFrame({"anchor": ["35 cm approach", "8 cm capture"], "trials with observed entry": [12, capture_anchor_count]}))
```

## What did we learn?

For these prescribed trajectories, **5 of 6 predictive attempts** enter the
capture sphere, versus **1 of 6 reactive attempts**. The gripper-frame view exposes
a larger lateral offset in the reactive examples. Event alignment makes the
approaches comparable even though their recording times differ.

The result follows from visible TAL operations: CSV ingestion → field assembly →
Pose registration → frame transformation → norm → reduction/grouping → event windows.
The raw-data generator performs none of those analysis steps.

The radius comparison is worked out above. The approach population stays fixed while the capture criterion changes, keeping misses visible. A change of anchor definition changes the population and therefore changes what the event-aligned comparison can establish.
